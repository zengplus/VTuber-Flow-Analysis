考点1：自定义函数 + 条件判断 + 循环
 
考试第1题：特殊累加计算
 
设计一个函数，用户可以通过键盘输入一个实数：
 
- 若该数值为 0 或 负值，则返回 0；
​
- 若该数值为 正值，则返回以下特殊累加结果：
 
示例说明：
 
1. 若输入：6.7
结果为：
1 + 22 + 333 + 4444 + 55555 + 666666
​
2. 若输入：4.2
结果为：
1 + 22 + 333 + 4444
 
（注：只取整数部分，按整数进行累加）

```python
def calc_special_sum():
    """
    考试真题：输入实数返回指定累加值
    规则：≤0返回0；>0取整数部分，计算1+22+333+...+n个n
    结合第一章input/print/循环/条件判断实现
    """
    try:
        a = float(input("num"))
    except ValueError:
        print("输入错误！请输入一个实数！")
        return

    if a <= 0:
        print(0)
        return

    n = int(a)
    total = 0
    for i in range(1, n + 1):
        # 生成每一个要加的数（比如i=2时，生成22）
        # 拆成3小步：
        # 1. str(i) → 把i变成文字（i=2→"2"）
        # 2. *i → 重复i次（"2"*2→"22"）
        # 3. int() → 变回数字（"22"→22）
        item = int(str(i) * i)
        total += item
        
    print(total)

calc_special_sum()
```

```python

def calc_special_sum():
    """
    考试真题：输入实数返回指定累加值
    规则：≤0返回0；>0取整数部分，计算1+22+333+...+n个n
    结合第一章input/print/循环/条件判断实现
    """
    try:
        num = float(input("请输入一个实数："))  # 第一章input函数+类型转换
    except ValueError:
        print("输入错误！请输入数字类型！")
        return 0
    # 第一章条件判断
    if num <= 0:
        print(f"输入值{num}≤0，返回结果：0")
        return 0
    n = int(num)
    total = 0
    # 第一章for循环
    for i in range(1, n+1):
        item = int(str(i) * i)
        total += item
    print(f"输入值{num}的整数部分为{n}，累加结果：1+22+...+{n}个{n} = {total}\n")
    return total
calc_special_sum()
```

考点2：辗转相除法
 
考试第2题：最大公约数 + 最小公倍数
 
从键盘输入两个正整数，使用辗转相除法计算并输出它们的最大公约数和最小公倍数。



```python
def calc_gcd_lcm():
    """
    考试真题：输入两个正整数，计算最大公约数（GCD）和最小公倍数（LCM）
    结合第一章input/循环/运算实现，辗转相除法核心
    """
calc_gcd_lcm()
```

```python
def calc_gcd_lcm():
    """
    try-except：异常处理，防止用户输入错误导致程序崩溃：
        map(int, ...)：将输入的字符串转换为整数，比如 12 18 转换为 12, 18。
        input(...)：接收用户输入的字符串，比如用户输入 12 18。
        split()：将字符串按空格分隔，比如 "12 18" 分隔为 "12", "18"。
    """
    try:
        a, b = map(int, input("请输入两个正整数，用空格分隔：").split())  # 第一章类型转换
    except (ValueError, IndexError):
        print("输入错误！请输入两个正整数！")
        return
    """
    检查用户输入的两个数是否是正整数（大于 0），如果不是，打印错误提示并结束函数。
    """
    if a <= 0 or b <= 0:
        print("输入错误！请输入正整数！")
        return
    """
    a_org, b_org = a, b：把用户输入的原始数值保存下来
    （因为后面循环会修改 a 和 b，计算最小公倍数需要用到原始值）。
    辗转相除法原理：用较大的数除以较小的数，再用除数除以余数，重复这个过程，
        直到余数为 0，此时的除数就是最大公约数。
        举个例子：输入 12 和 18
        初始：a=12, b=18 → 进入循环（b≠0）
        第一次循环：a, b = 18, 12%18=12 → a=18, b=12
        第二次循环：a, b = 12, 18%12=6 → a=12, b=6
        第三次循环：a, b = 6, 12%6=0 → a=6, b=0
        循环结束（b=0），此时 a=6 就是最大公约数。
        gcd = a：循环结束后，a 的值就是最大公约数，赋值给 gcd 变量。
    """
    a_org, b_org = a, b
    # 第一章while循环（辗转相除法核心）
    while b != 0:
        a, b = b, a % b
    gcd = a
    lcm = int(a_org * b_org / gcd)
    print(f"最大公约数：{gcd}，最小公倍数：{lcm}\n")
    return gcd, lcm
calc_gcd_lcm()
```
